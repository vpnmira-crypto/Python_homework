У завданнях цього розділу працюємо з каталогом онлайн-курсів. Кожен курс — словник з назвою, категорією, ціною (грн), кількістю студентів, рейтингом і тривалістю в годинах. **Спершу запусти клітинку з даними.**

```python
COURSES = [
    {"title": "SQL Basics", "category": "data", "price": 1200, "students": 340, "rating": 4.7, "hours": 12},
    {"title": "Python for Analysts", "category": "data", "price": 2400, "students": 515, "rating": 4.8, "hours": 30},
    {"title": "Excel Pro", "category": "office", "price": 900, "students": 610, "rating": 4.4, "hours": 10},
    {"title": "Tableau Dashboards", "category": "bi", "price": 1800, "students": 220, "rating": 4.6, "hours": 16},
    {"title": "Statistics 101", "category": "data", "price": 1500, "students": 280, "rating": 4.3, "hours": 20},
    {"title": "Power BI Start", "category": "bi", "price": 1600, "students": 390, "rating": 4.5, "hours": 14},
    {"title": "Presentation Skills", "category": "office", "price": 700, "students": 150, "rating": 4.1, "hours": 6},
    {"title": "Machine Learning Intro", "category": "data", "price": 3200, "students": 205, "rating": 4.9, "hours": 40},
]
```

1. Відсортуй курси за кількістю студентів — від найпопулярнішого до найменш популярного — і виведи назву та кількість студентів кожного курсу.  
Потім знайди та виведи назву найкоротшого курсу (за кількістю годин) і назву найдорожчого курсу

```python

# Сортуємо курси за кількістю студентів від більшої до меншої
sorted_courses = sorted(COURSES, key=lambda course: course["students"], reverse=True)

# Виводимо назву курсу та кількість студентів
for course in sorted_courses:
    print(course["title"], "-", course["students"])

# Знаходимо найкоротший курс за кількістю годин
shortest_course = min(COURSES, key=lambda course: course["hours"])

# Знаходимо найдорожчий курс за ціною
most_expensive_course = max(COURSES, key=lambda course: course["price"])

# Виводимо результати
print("Найкоротший курс:", shortest_course["title"])
print("Найдорожчий курс:", most_expensive_course["title"])

```

2. Відбери курси з категорії `"data"` з рейтингом 4.5 і вище. Для відібраних курсів отримай:
- кількість таких курсів;
- список їхніх назв великими літерами;
- список цін зі знижкою 15%, округлених до цілих.

Розвʼяжи завдання за допомогою `filter` і `map`

```python

# 1. Відбираємо курси категорії "data" з рейтингом 4.5 і вище
filtered_courses = list(filter(
    lambda course: course["category"] == "data" and course["rating"] >= 4.5,
    COURSES
))

# 2. Рахуємо кількість відібраних курсів
print("Кількість курсів:", len(filtered_courses))

# 3. Отримуємо назви курсів великими літерами
course_names = list(map(
    lambda course: course["title"].upper(),
    filtered_courses
))
print("Назви курсів:", course_names)

# 4. Розраховуємо ціни зі знижкою 15%
discounted_prices = list(map(
    lambda course: round(course["price"] * 0.85),
    filtered_courses
))
print("Ціни зі знижкою:", discounted_prices)
```

3. Напиши функцію `top_by(items, field, n=3, condition=None)` з docstring, яка повертає `n` елементів з найбільшим значенням поля `field`. Якщо передано `condition` — функцію-умову, — спершу залиш лише ті елементи, для яких вона повертає `True`.

Виклич функцію й виведи назви курсів:
- 3 курси з найвищим рейтингом;
- 2 найпопулярніші курси (за кількістю студентів) серед курсів категорії `"bi"`;
- 1 найдовший курс (за кількістю годин)

```python
# Створюємо функцію для пошуку найкращих курсів
def top_by(items, field, n=3, condition=None):
    """
    Повертає n елементів із найбільшими значеннями поля field.
    Якщо задано condition, спочатку відбирає елементи за умовою.
    """

# Якщо є умова, відбираємо потрібні курси
    if condition is not None:
        items = list(filter(condition, items))

# Сортуємо курси від найбільшого значення до найменшого
    sorted_items = sorted(items, key=lambda item: item[field], reverse=True)

# Повертаємо потрібну кількість курсів
    return sorted_items[:n]

# 1. Знаходимо три курси з найвищим рейтингом
top_rating = top_by(COURSES, "rating", 3)

print("Три курси з найвищим рейтингом:")
for course in top_rating:
    print(course["title"])

# 2. Знаходимо два найпопулярніші курси категорії "bi"
top_bi = top_by(
    COURSES,
    "students",
    2,
    condition=lambda course: course["category"] == "bi"
)

print("\nДва найпопулярніші BI-курси:")
for course in top_bi:
    print(course["title"])

# 3. Знаходимо один найдовший курс
longest_course = top_by(COURSES, "hours", 1)

print("\nНайдовший курс:")
for course in longest_course:
    print(course["title"])
```

4. Служба доставки бере 35 грн за кожен **розпочатий** кілограм ваги посилки: посилка вагою 1.2 кг оплачується як 2 кг.  
Дано ваги посилок у кілограмах: `weights = [0.4, 2.0, 3.7, 1.2, 12.5, 0.9]`.

Виведи вартість доставки кожної посилки та загальну вартість. Окремо виведи загальну вагу всіх посилок, округлену **вниз** до цілих кілограмів

```python

# Підключаємо модуль math для округлення
import math

# Список ваг посилок у кілограмах
weights = [0.4, 2.0, 3.7, 1.2, 12.5, 0.9]

# Вартість доставки за 1 кг
price_per_kg = 35

# Змінна для загальної вартості доставки
total_cost = 0

# Розраховуємо вартість кожної посилки
for weight in weights:
    paid_weight = math.ceil(weight)
    cost = paid_weight * price_per_kg
    total_cost += cost
    print(f"Вага: {weight} кг — доставка: {cost} грн")

# Обчислюємо загальну вагу всіх посилок
total_weight = math.floor(sum(weights))

# Виводимо підсумки
print("Загальна вартість доставки:", total_cost, "грн")
print("Загальна вага (округлена вниз):", total_weight, "кг")
```

5. Дано час відповіді служби підтримки на звернення клієнтів (у хвилинах) за два тижні:

`week_1 = [12, 15, 9, 15, 22, 18, 15, 11, 14, 16]`  
`week_2 = [5, 30, 8, 27, 8, 25, 9, 31, 8, 28]`

За допомогою модуля `statistics` для кожного тижня виведи середнє, медіану, моду та стандартне відхилення (середнє й стандартне відхилення округли до 2 знаків).  
Виведи, у який тиждень служба працювала **стабільніше** (час відповіді менше коливався), і поясни в коментарі, на якому показнику ґрунтується висновок

```python

# Підключаємо модуль statistics
import statistics

# Час відповіді за перший тиждень
week_1 = [12, 15, 9, 15, 22, 18, 15, 11, 14, 16]

# Час відповіді за другий тиждень
week_2 = [5, 30, 8, 27, 8, 25, 9, 31, 8, 28]

# Розраховуємо показники першого тижня
print("Перший тиждень:")
print("Середнє:", round(statistics.mean(week_1), 2))
print("Медіана:", statistics.median(week_1))
print("Мода:", statistics.mode(week_1))
print("Стандартне відхилення:", round(statistics.stdev(week_1), 2))

# Розраховуємо показники другого тижня
print("\nДругий тиждень:")
print("Середнє:", round(statistics.mean(week_2), 2))
print("Медіана:", statistics.median(week_2))
print("Мода:", statistics.mode(week_2))
print("Стандартне відхилення:", round(statistics.stdev(week_2), 2))

# Порівнюємо стабільність роботи
if statistics.stdev(week_1) < statistics.stdev(week_2):
    print("\nПерший тиждень був стабільнішим.")
else:
    print("\nДругий тиждень був стабільнішим.")

# Висновок: стабільність визначаємо за стандартним відхиленням.
# Чим менше стандартне відхилення, тим менше коливається час відповіді.
```

6. Згенеруй тестові дані для 10 замовлень. Встанови `seed` рівним 2026. Для кожного замовлення спочатку згенеруй суму — ціле число від 100 до 5000 включно, — а потім випадково обери спосіб оплати зі списку `["card", "cash", "apple pay"]`. Збережи замовлення як список кортежів `(сума, спосіб оплати)`.

Виведи:
- усі згенеровані замовлення;
- кількість замовлень кожним способом оплати;
- частку замовлень на суму понад 2000 грн у відсотках;
- 2 різних випадкових замовлення для перевірки якості, вибрані з уже згенерованого списку

```python

# Підключаємо модуль random
import random

# Встановлюємо seed для відтворюваних результатів
random.seed(2026)

# Способи оплати
payment_methods = ["card", "cash", "apple_pay"]

# Створюємо порожній список замовлень
orders = []

# Генеруємо 10 випадкових замовлень
for i in range(10):
    amount = random.randint(100, 5000)
    payment = random.choice(payment_methods)
    orders.append((amount, payment))

# Виводимо всі замовлення
print("Усі замовлення:")
for order in orders:
    print(order)

# Рахуємо кількість замовлень за способом оплати
print("\nКількість замовлень за способом оплати:")
for method in payment_methods:
    count = sum(1 for order in orders if order[1] == method)
    print(method, "-", count)

# Рахуємо частку замовлень понад 2000 грн
expensive_orders = sum(1 for order in orders if order[0] > 2000)
percentage = expensive_orders / len(orders) * 100

print("\nЧастка замовлень понад 2000 грн:", round(percentage, 2), "%")

# Вибираємо 2 різні випадкові замовлення для перевірки
sample_orders = random.sample(orders, 2)

print("\nДва замовлення для перевірки:")
for order in sample_orders:
    print(order)
```

7. **Аналіз підписок.** Дано підписки клієнтів у форматі `(клієнт, дата початку, тривалість у днях)` і дату, на яку робиться звіт:

`TODAY = "2026-10-21"`  
`subscriptions = [("anna", "2026-09-01", 30), ("bohdan", "2026-10-01", 30), ("iryna", "2026-08-15", 90), ("oleh", "2026-10-10", 14), ("sofia", "2026-07-01", 365), ("taras", "2026-10-15", 7)]`

Для кожної підписки визнач:
- дату закінчення у форматі `ДД.ММ.РРРР`;
- день тижня, на який припадає дата закінчення (англійською, наприклад `Saturday`);
- скільки днів залишилося до закінчення від `TODAY` (для завершених підписок — відʼємне число);
- статус: `"expired"` — підписка вже закінчилася, `"expires soon"` — залишилося 7 днів або менше, `"active"` — в інших випадках.

Виведи результат таблицею за допомогою пакета `tabulate`. Наприкінці виведи кількість підписок кожного статусу та медіану кількості днів, що залишилися, для активних підписок

```python

# Підключаємо необхідні модулі
from datetime import datetime, timedelta
from collections import Counter
from statistics import median
from tabulate import tabulate

# Дата, на яку робимо звіт
TODAY = "2026-10-21"

# Список підписок: клієнт, дата початку, тривалість у днях
subscriptions = [
    ("anna", "2026-09-01", 30),
    ("bohdan", "2026-10-01", 30),
    ("iryna", "2026-08-15", 90),
    ("oleh", "2026-10-10", 14),
    ("sofia", "2026-07-01", 365),
    ("taras", "2026-10-15", 7)
]

# Перетворюємо текстову дату на справжню дату
today = datetime.strptime(TODAY, "%Y-%m-%d").date()

# Створюємо порожній список для результатів
results = []

# Аналізуємо кожну підписку
for name, start_date, duration in subscriptions:

    # Перетворюємо дату початку
    start = datetime.strptime(start_date, "%Y-%m-%d").date()

    # Обчислюємо дату закінчення
    end = start + timedelta(days=duration)

    # Рахуємо, скільки днів залишилося
    days_left = (end - today).days

    # Визначаємо статус підписки
    if days_left < 0:
        status = "expired"
    elif days_left <= 7:
        status = "expires_soon"
    else:
        status = "active"

    # Додаємо результат до списку
    results.append([
        name,
        end.strftime("%d.%m.%Y"),
        end.strftime("%A"),
        days_left,
        status
    ])

# Виводимо результати у вигляді таблиці
print(tabulate(
    results,
    headers=["Клієнт", "Дата закінчення", "День тижня", "Днів залишилось", "Статус"],
    tablefmt="grid"
))

# Рахуємо кількість підписок кожного статусу
status_counts = Counter(row[4] for row in results)

print("\nКількість підписок за статусами:")
for status in ["expired", "expires_soon", "active"]:
    print(status, "-", status_counts[status])

# Знаходимо дні, що залишилися для активних підписок
active_days = [row[3] for row in results if row[4] == "active"]

# Обчислюємо медіану
if active_days:
    print("\nМедіана днів для активних підписок:", median(active_days))
else:
    print("\nАктивних підписок немає")
```
