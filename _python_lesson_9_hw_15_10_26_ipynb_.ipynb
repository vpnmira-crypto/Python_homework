## Завдання 1. Перевірка та очищення даних

Переглянь усі поїздки та розподіли їх:
- скасовані поїздки;
- **некоректні** завершені поїздки: відстань дорівнює 0 або менше — причина `"invalid distance"`; оцінка не є цілим числом від 1 до 5 — причина `"invalid rating"`. Якщо в поїздці кілька проблем, вказуй лише першу з них у цьому порядку;
- **коректні** завершені поїздки — збережи їх в окремий список `valid_rides`, з ним ти працюватимеш далі.

Виведи загальну кількість поїздок, кількість скасованих, кількість коректних, а також номер і причину для кожної некоректної поїздки

```python

# Створюємо три порожні списки
cancelled_rides = []  # Скасовані поїздки
invalid_rides = []    # Некоректні поїздки
valid_rides = []      # Коректні поїздки

# Перебираємо всі поїздки
for ride in rides:

    # Перевіряємо, чи скасована поїздка
    if ride["status"] == "cancelled":
        cancelled_rides.append(ride)

    # Перевіряємо відстань
    elif ride["distance"] <= 0:
        invalid_rides.append(
            (ride["id"], "invalid distance")
        )

    # Перевіряємо рейтинг
    elif type(ride["rating"]) is not int or not 1 <= ride["rating"] <= 5:
        invalid_rides.append(
            (ride["id"], "invalid rating")
        )

    # Якщо помилок немає
    else:
        valid_rides.append(ride)

# Виводимо результати
print("Всього поїздок:", len(rides))
print("Скасованих:", len(cancelled_rides))
print("Коректних:", len(valid_rides))
print("Некоректних:", len(invalid_rides))

# Виводимо причини помилок
for ride_id, reason in invalid_rides:
    print("Поїздка:", ride_id, "| Причина:", reason)

```

## Завдання 2. Розрахунок вартості поїздок

Для кожної поїздки з `valid_rides` розрахуй вартість за тарифом:
- посадка — 50 грн плюс 12 грн за кожен кілометр;
- якщо поїздка тривала понад 30 хвилин — додатково 3 грн за кожну хвилину понад 30;
- нічний тариф (поїздка почалася з 22:00 до 05:59 включно) — уся вартість збільшується на 25%;
- мінімальна вартість поїздки — 80 грн: якщо розрахована сума менша, пасажир платить 80 грн.

Округли вартість до 2 знаків і додай її до словника кожної поїздки з ключем `"price"`. Виведи для кожної поїздки номер, водія, відстань і вартість

```python

# Перебираємо всі коректні поїздки
for ride in valid_rides:

    # 1. Посадка 50 грн + 12 грн за кожен кілометр
    price = 50 + ride["distance"] * 12

    # 2. Доплата за час понад 30 хвилин
    if ride["duration"] > 30:
        extra_minutes = ride["duration"] - 30
        price = price + extra_minutes * 3

    # 3. Нічний тариф з 22:00 до 05:59
    if ride["hour"] >= 22 or ride["hour"] < 6:
        price = price * 1.25

    # 4. Мінімальна вартість 80 грн
    if price < 80:
        price = 80

    # 5. Округлюємо до двох знаків
    price = round(price, 2)

    # 6. Записуємо вартість у словник поїздки
    ride["price"] = price

    # 7. Виводимо результат
    print(
        "Поїздка:", ride["id"],
        "| Водій:", ride["driver"],
        "| Відстань:", ride["distance"], "км",
        "| Вартість:", ride["price"], "грн"
    )
```

## Завдання 3. Статистика по водіях

На основі `valid_rides` створи словник `driver_stats`, у якому для кожного водія зберігаються:
- кількість поїздок;
- виручка (сума вартостей його поїздок, округлена до 2 знаків);
- середній рейтинг, округлений до 1 знака.

Виведи статистику кожного водія в одному рядку

```python

# Створюємо порожній словник для статистики водіїв
driver_stats = {}

# Перебираємо всі коректні поїздки
for ride in valid_rides:

    # Отримуємо ім'я водія
    driver = ride["driver"]

    # Якщо водія ще немає у словнику
    if driver not in driver_stats:

        # Створюємо для нього новий словник
        driver_stats[driver] = {
            "count": 0,        # Кількість поїздок
            "revenue": 0,      # Загальна виручка
            "rating_sum": 0    # Сума всіх рейтингів
        }

    # Додаємо одну поїздку
    driver_stats[driver]["count"] += 1

    # Додаємо вартість поїздки до виручки
    driver_stats[driver]["revenue"] += ride["price"]

    # Додаємо рейтинг поїздки
    driver_stats[driver]["rating_sum"] += ride["rating"]

# Обчислюємо підсумкову статистику
for driver, stats in driver_stats.items():

    # Округлюємо виручку до 2 знаків
    stats["revenue"] = round(stats["revenue"], 2)

    # Обчислюємо середній рейтинг
    stats["avg_rating"] = round(
        stats["rating_sum"] / stats["count"], 1
    )

    # Видаляємо допоміжну суму рейтингів
    del stats["rating_sum"]

    # Виводимо результат
    print(
        "Водій:", driver,
        "| Поїздок:", stats["count"],
        "| Виручка:", stats["revenue"], "грн",
        "| Середній рейтинг:", stats["avg_rating"]
    )
```

## Завдання 4. Статуси та бонуси водіїв

На основі `driver_stats` визнач для кожного водія статус і бонус (середній рейтинг береться округленим до 1 знака):
- рейтинг від 4.8 **і** щонайменше 3 поїздки — статус `"Gold"`, бонус 10% від виручки;
- рейтинг від 4.5 — статус `"Silver"`, бонус 5% від виручки;
- рейтинг нижче 4.0 — статус `"Warning"`, бонусу немає;
- в інших випадках — статус `"Standard"`, бонусу немає.

Округли бонус до 2 знаків і додай статус та бонус до статистики водія. Виведи імʼя, статус і бонус кожного водія, а наприкінці — загальну суму бонусів

```python

# Загальна сума бонусів усіх водіїв
total_bonus = 0

# Перебираємо водіїв та їхню статистику
for driver, stats in driver_stats.items():

    # Отримуємо середній рейтинг і кількість поїздок
    rating = stats["avg_rating"]
    count = stats["count"]
    revenue = stats["revenue"]

    # Gold: рейтинг від 4.8 і мінімум 3 поїздки
    if rating >= 4.8 and count >= 3:
        status = "Gold"
        bonus = revenue * 0.10

    # Silver: рейтинг від 4.5
    elif rating >= 4.5:
        status = "Silver"
        bonus = revenue * 0.05

    # Warning: рейтинг нижче 4.0
    elif rating < 4.0:
        status = "Warning"
        bonus = 0

    # Standard: усі інші випадки
    else:
        status = "Standard"
        bonus = 0

    # Округлюємо бонус до 2 знаків
    bonus = round(bonus, 2)

    # Додаємо статус і бонус до словника водія
    stats["status"] = status
    stats["bonus"] = bonus

    # Додаємо бонус до загальної суми
    total_bonus += bonus

    # Виводимо результат для кожного водія
    print(
        "Водій:", driver,
        "| Статус:", status,
        "| Бонус:", bonus, "грн"
    )

# Виводимо загальну суму бонусів
total_bonus = round(total_bonus, 2)
print("Загальна сума бонусів:", total_bonus, "грн")
```

## Завдання 5. Підсумковий звіт за день

Виведи звіт, у якому кожен показник займає окремий рядок:
- загальна виручка за день;
- кількість коректних поїздок і середня вартість поїздки (округли до 2 знаків);
- частка скасованих поїздок від усіх поїздок у журналі, % (округли до 1 знака);
- частка поїздок, оплачених карткою, серед коректних поїздок, % (округли до 1 знака);
- найзавантаженіша година — година, на яку припало найбільше коректних поїздок, і кількість поїздок у цю годину;
- найдовша поїздка: номер, водій і відстань;
- водій з найбільшою виручкою;
- загальна сума бонусів водіям.

Не використовуй функції `max()` і `min()`

```python

# 1. Загальна виручка за день
total_revenue = 0

# Кількість поїздок, оплачених карткою
card_count = 0

# Словник для підрахунку поїздок за годинами
hour_counts = {}

# Змінна для найдовшої поїздки
longest_ride = None

# Перебираємо всі коректні поїздки
for ride in valid_rides:

    # Додаємо вартість до загальної виручки
    total_revenue += ride["price"]

    # Перевіряємо оплату карткою
    if ride["payment"] == "card":
        card_count += 1

    # Отримуємо годину поїздки
    hour = ride["hour"]

    # Рахуємо кількість поїздок за годинами
    if hour not in hour_counts:
        hour_counts[hour] = 0

    hour_counts[hour] += 1

    # Шукаємо найдовшу поїздку
    if longest_ride is None or ride["distance"] > longest_ride["distance"]:
        longest_ride = ride

# 2. Кількість коректних поїздок
valid_count = len(valid_rides)

# 3. Середня вартість поїздки
average_price = round(total_revenue / valid_count, 2)

# 4. Частка скасованих поїздок
cancelled_percent = round(
    len(cancelled_rides) / len(rides) * 100, 1
)

# 5. Частка оплат карткою
card_percent = round(
    card_count / valid_count * 100, 1
)

# 6. Шукаємо найзавантаженішу годину
busiest_hour = None
busiest_count = 0

for hour, count in hour_counts.items():

    if count > busiest_count:
        busiest_count = count
        busiest_hour = hour

# 7. Шукаємо водія з найбільшою виручкою
top_driver = None
top_revenue = 0

for driver, stats in driver_stats.items():

    if stats["revenue"] > top_revenue:
        top_revenue = stats["revenue"]
        top_driver = driver

# 8. Виводимо підсумковий звіт
print("ПІДСУМКОВИЙ ЗВІТ ЗА ДЕНЬ")
print("---------------------------")

print("Загальна виручка:", round(total_revenue, 2), "грн")

print("Кількість коректних поїздок:", valid_count)

print("Середня вартість поїздки:", average_price, "грн")

print("Частка скасованих поїздок:", cancelled_percent, "%")

print("Частка оплат карткою:", card_percent, "%")

print("Найзавантаженіша година:", busiest_hour)
print("Кількість поїздок у цю годину:", busiest_count)

print("Найдовша поїздка:", longest_ride["id"])
print("Водій:", longest_ride["driver"])
print("Відстань:", longest_ride["distance"], "км")

print("Водій з найбільшою виручкою:", top_driver)
print("Найбільша виручка:", top_revenue, "грн")

print("Загальна сума бонусів:", total_bonus, "грн")
```
