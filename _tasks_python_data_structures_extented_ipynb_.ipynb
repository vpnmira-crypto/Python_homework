## 📘 Вступ: колекції в Python

До цього ми працювали зі змінними, які зберігають **одне** значення: число, рядок, `True`/`False`. Але в аналітиці даних ми майже завжди маємо справу з **набором** значень: продажі за кожен день місяця, міста клієнтів, унікальні категорії товарів, картка клієнта з багатьма полями.

Для цього в Python є **колекції** — типи даних, які зберігають багато значень в одній змінній. Сьогодні розберемо чотири основні:

| Тип | Як виглядає | Порядок / індекси | Можна змінювати? | Дублікати | Приклад з аналітики |
|---|---|---|---|---|---|
| `list` — список | `[1, 2, 3]` | ✅ є індекси | ✅ так | ✅ можна | продажі за кожен день |
| `tuple` — кортеж | `(1, 2, 3)` | ✅ є індекси | ❌ ні | ✅ можна | координати точки, дата (рік, місяць, день) |
| `set` — множина | `{1, 2, 3}` | ❌ порядку немає | ✅ так | ❌ лише унікальні | унікальні міста клієнтів |
| `dict` — словник | `{"a": 1, "b": 2}` | доступ за **ключем** | ✅ так | ключі — ні, значення — так | картка клієнта: імʼя, вік, місто |

> 💡 **Змінюваний (mutable)** тип можна змінити «на місці» після створення: додати, видалити чи замінити елемент. **Незмінюваний (immutable)** — ні: будь-яка «зміна» насправді створює новий обʼєкт. Рядки, з якими ви працювали вчора, — незмінювані.

## 1. Lists (Списки) — 10 завдань

### 📘 Теорія: що таке список

**Список (`list`)** — впорядкована змінювана колекція елементів. Це найуживаніша колекція в Python.

- Створюється квадратними дужками, елементи пишуться через кому.
- Може містити елементи різних типів (хоча на практиці зазвичай вони однотипні).
- Елементи можуть повторюватися.
- Порожній список: `[]` або `list()`.
- Функція `list()` перетворює на список інші колекції — наприклад, рядок розбивається на окремі символи.

````python
# Створення списків
prices = [120, 85, 340, 85, 99]            # числа (85 повторюється — це нормально)
cities = ["Kyiv", "Lviv", "Odesa"]         # рядки
mixed = ["Anna", 28, True, 1.75]           # різні типи — так теж можна
empty = []                                 # порожній список

print(prices)
print(type(cities))       # <class 'list'>
print(list("data"))       # ['d', 'a', 't', 'a']
````

### 📘 Індекси та зрізи

Списки індексуються **так само, як рядки** — це ви вже знаєте з минулого заняття:

- нумерація починається з **0**;
- відʼємні індекси рахують з кінця: `-1` — останній елемент, `-2` — передостанній;
- зріз `lst[start:stop:step]` повертає **новий список**, елемент з індексом `stop` не включається;
- будь-який параметр зрізу можна пропустити: `lst[:3]`, `lst[2:]`, `lst[::2]`;
- `lst[::-1]` — копія списку у зворотному порядку.

```
cities = ["Kyiv", "Lviv", "Odesa", "Dnipro", "Kharkiv"]
індекс:     0       1       2        3         4
з кінця:   -5      -4      -3       -2        -1
```

````python
cities = ["Kyiv", "Lviv", "Odesa", "Dnipro", "Kharkiv"]

print(cities[0])      # Kyiv     — перший
print(cities[-1])     # Kharkiv  — останній
print(cities[1:3])    # ['Lviv', 'Odesa']  — індекси 1 і 2 (3 не включається)
print(cities[:2])     # ['Kyiv', 'Lviv']   — перші два
print(cities[::2])    # ['Kyiv', 'Odesa', 'Kharkiv'] — кожен другий
print(cities[::-1])   # ['Kharkiv', 'Dnipro', 'Odesa', 'Lviv', 'Kyiv']

# cities[10]  →  IndexError: list index out of range (такого індексу немає)
````

### 📘 Зміна списку

На відміну від рядків, список **можна змінювати**.

| Дія | Синтаксис | Що робить |
|---|---|---|
| Замінити елемент | `lst[i] = value` | записує нове значення за індексом `i` |
| Додати в кінець | `lst.append(x)` | додає **один** елемент у кінець |
| Вставити | `lst.insert(i, x)` | вставляє `x` на позицію `i`, решта зсувається вправо |
| Додати кілька | `lst.extend([x, y])` | додає всі елементи іншої колекції |
| Видалити за значенням | `lst.remove(x)` | видаляє **перше** входження `x` (помилка, якщо його немає) |
| Видалити за індексом | `lst.pop(i)` | видаляє елемент з індексом `i` і **повертає** його; без аргументу — останній |
| Видалити за індексом | `del lst[i]` | просто видаляє елемент |
| Очистити | `lst.clear()` | робить список порожнім |

> ⚠️ Методи `append`, `insert`, `remove`, `sort`, `reverse` змінюють список **на місці** й нічого не повертають (точніше, повертають `None`). Тому запис `lst = lst.append(x)` — помилка: у `lst` опиниться `None`.

````python
sales = [100, 200, 300]

sales[0] = 150            # заміна
print(sales)              # [150, 200, 300]

sales.append(400)         # у кінець
print(sales)              # [150, 200, 300, 400]

sales.insert(1, 175)      # на позицію з індексом 1
print(sales)              # [150, 175, 200, 300, 400]

sales.extend([500, 600])  # кілька елементів одразу
print(sales)              # [150, 175, 200, 300, 400, 500, 600]

sales.remove(300)         # за значенням
print(sales)              # [150, 175, 200, 400, 500, 600]

last = sales.pop()        # видаляє останній і повертає його
print(last, sales)        # 600 [150, 175, 200, 400, 500]

del sales[0]              # за індексом
print(sales)              # [175, 200, 400, 500]
````

### 📘 Корисні функції та методи

| Що потрібно | Синтаксис | Результат |
|---|---|---|
| Довжина | `len(lst)` | кількість елементів |
| Чи є елемент | `x in lst` | `True` / `False` |
| Скільки разів зустрічається | `lst.count(x)` | число |
| Позиція елемента | `lst.index(x)` | індекс першого входження |
| Сума, мінімум, максимум | `sum(lst)`, `min(lst)`, `max(lst)` | для числових списків |
| Відсортована копія | `sorted(lst)` | **новий** список, оригінал не змінюється |
| Сортування на місці | `lst.sort()` | змінює сам список |
| Розвернути на місці | `lst.reverse()` | змінює сам список |
| Розвернута копія | `lst[::-1]` | **новий** список, оригінал не змінюється |

**Функція `range(start, stop, step)`** генерує послідовність цілих чисел; `stop` не включається, як і в зрізах. Щоб побачити ці числа, перетворіть результат на список: `list(range(1, 6))` → `[1, 2, 3, 4, 5]`.

````python
scores = [7, 3, 9, 3, 5, 3]

print(len(scores))        # 6
print(9 in scores)        # True
print(scores.count(3))    # 3
print(scores.index(9))    # 2
print(sum(scores), min(scores), max(scores))   # 30 3 9

print(sorted(scores))     # [3, 3, 3, 5, 7, 9] — нова копія
print(scores)             # [7, 3, 9, 3, 5, 3] — оригінал без змін

scores.sort()             # сортуємо сам список
print(scores)             # [3, 3, 3, 5, 7, 9]

print(list(range(1, 6)))       # [1, 2, 3, 4, 5]
print(list(range(0, 20, 5)))   # [0, 5, 10, 15]
````

### ⚠️ Пастка: копіювання списку

`b = a` **не** створює копію — обидві змінні вказують на **той самий** список. Змінюєте один — змінюється й «інший». Незалежну копію дають `a.copy()`, `list(a)` або `a[:]`.

````python
a = [1, 2, 3]
b = a           # той самий список!
c = a.copy()    # незалежна копія

a.append(4)
print(b)        # [1, 2, 3, 4] — змінився разом з a
print(c)        # [1, 2, 3]
````

## 2. Tuples (Кортежі) — 10 завдань

### 📘 Теорія: що таке кортеж

**Кортеж (`tuple`)** — впорядкована, але **незмінювана** колекція. Після створення в ньому не можна замінити, додати чи видалити елементи.

Навіщо він, якщо є списки?
- **захист даних**: значення, які не мають змінюватися (координати, RGB-колір, дата як рік-місяць-день);
- кортеж трохи швидший і займає менше памʼяті;
- кортеж, на відміну від списку, може бути **ключем словника** або елементом множини;
- функції часто повертають кілька значень саме у вигляді кортежу.

**Створення:**
- круглі дужки: `(1, 2, 3)`; дужки можна навіть пропустити: `1, 2, 3`;
- порожній кортеж: `()` або `tuple()`;
- `tuple(...)` перетворює список чи рядок на кортеж.

> ⚠️ **Кортеж з одного елемента** потребує коми: `("red",)`. Без коми `("red")` — це просто рядок у дужках! Кортеж утворює саме **кома**, а не дужки.

````python
point = (50.45, 30.52)          # координати
rgb = (255, 128, 0)             # колір
date = 2024, 9, 1               # дужки не обовʼязкові
print(type(date))               # <class 'tuple'>

not_tuple = ("red")
real_tuple = ("red",)
print(type(not_tuple))          # <class 'str'>
print(type(real_tuple))         # <class 'tuple'>

print(tuple([1, 2, 3]))         # (1, 2, 3)
````

### 📘 Робота з кортежами

Усе, що **не змінює** колекцію, працює так само, як у списків:

| Що потрібно | Синтаксис |
|---|---|
| Індекс / зріз | `t[0]`, `t[-1]`, `t[1:3]` |
| Довжина | `len(t)` |
| Перевірка наявності | `x in t` |
| Підрахунок / пошук позиції | `t.count(x)`, `t.index(x)` |
| Сума, мінімум, максимум | `sum(t)`, `min(t)`, `max(t)` |
| Обʼєднання | `t1 + t2` → **новий** кортеж |
| Повторення | `t * 3` → **новий** кортеж |

У кортежа лише два методи: `count()` та `index()`. Методів `append`, `remove`, `sort` немає.

**Як «змінити» кортеж?** Сам кортеж — ніяк, але можна створити новий і записати його в ту саму змінну:
1. через обʼєднання: `t = t + (new_element,)` — зверніть увагу на кому;
2. через перетворення: `list(t)` → змінити список → `tuple(...)`.

````python
sizes = ("S", "M", "L")

print(sizes[1])           # M
print(len(sizes))         # 3
print("XL" in sizes)      # False

# sizes[0] = "XS"   →  TypeError: 'tuple' object does not support item assignment

more = sizes + ("XL", "XXL")     # новий кортеж
print(more)               # ('S', 'M', 'L', 'XL', 'XXL')

temp = list(sizes)        # кортеж → список
temp.insert(0, "XS")      # змінюємо список
sizes = tuple(temp)       # список → кортеж
print(sizes)              # ('XS', 'S', 'M', 'L')
````

### 📘 Розпакування кортежу

Елементи кортежу можна одразу «розкласти» по окремих змінних. Кількість змінних має збігатися з кількістю елементів. Це зручно, коли кортеж описує один запис: місто та населення, координати x та y тощо.

````python
city, population = ("Lviv", 717000)
print(city)         # Lviv
print(population)   # 717000
````

## 3. Sets (Множини) — 10 завдань

### 📘 Теорія: що таке множина

**Множина (`set`)** — **невпорядкована** колекція **унікальних** елементів, як множина в математиці.

- Дублікати автоматично відкидаються — це головна «суперсила» множини. В аналітиці: «скільки в нас унікальних клієнтів?», «які категорії товарів взагалі трапляються?».
- Порядку немає, тому **немає й індексів**: `s[0]` дасть помилку. Порядок виведення може відрізнятися від порядку, в якому ви записали елементи.
- Множина змінювана, але її елементи мають бути незмінюваними: числа, рядки, кортежі (списки — ні).
- Перевірка `x in s` у множині працює дуже швидко, навіть на мільйонах елементів.

**Створення:**
- фігурні дужки: `{1, 2, 3}`;
- функція `set()` з іншої колекції: `set([1, 1, 2])` → `{1, 2}`; з рядка — множина унікальних символів;
- ⚠️ порожня множина — **тільки** `set()`. Запис `{}` створює порожній **словник**!

````python
regions = {"North", "South", "East", "South", "North"}
print(regions)            # 3 елементи — дублікати зникли (порядок може бути будь-яким)
print(len(regions))       # 3

orders = [101, 102, 101, 103, 102, 101]
unique_orders = set(orders)
print(unique_orders)      # {101, 102, 103}
print(len(unique_orders)) # 3 — кількість унікальних значень

print(set("hello"))       # {'h', 'e', 'l', 'o'} — у довільному порядку

print(type({}))           # <class 'dict'>  ← не множина!
print(type(set()))        # <class 'set'>
````

### 📘 Зміна множини

| Дія | Синтаксис | Примітка |
|---|---|---|
| Додати елемент | `s.add(x)` | якщо `x` вже є — нічого не станеться |
| Додати кілька | `s.update([x, y])` | |
| Видалити | `s.remove(x)` | помилка `KeyError`, якщо `x` немає |
| Видалити «безпечно» | `s.discard(x)` | без помилки, якщо `x` немає |
| Видалити довільний | `s.pop()` | видаляє і повертає якийсь елемент |
| Очистити | `s.clear()` | залишається порожня множина `set()` |

````python
tags = {"sale", "new"}

tags.add("top")
tags.add("sale")          # вже є — дубліката не буде
print(len(tags))          # 3

tags.update(["hit", "eco"])
print(len(tags))          # 5

tags.remove("new")
tags.discard("absent")    # такого елемента немає, але й помилки немає
print(tags)

tags.clear()
print(tags)               # set()
````

### 📘 Операції над множинами

Найсильніша сторона множин — операції з математики. Кожну можна записати методом або оператором:

| Операція | Метод | Оператор | Результат |
|---|---|---|---|
| Обʼєднання (union) | `a.union(b)` | <code>a &#124; b</code> | усі елементи з обох множин |
| Перетин (intersection) | `a.intersection(b)` | `a & b` | тільки спільні елементи |
| Різниця (difference) | `a.difference(b)` | `a - b` | є в `a`, але немає в `b` |
| Симетрична різниця | `a.symmetric_difference(b)` | `a ^ b` | є лише в одній із двох множин |
| Підмножина | `a.issubset(b)` | `a <= b` | `True`, якщо всі елементи `a` є в `b` |
| Надмножина | `a.issuperset(b)` | `a >= b` | `True`, якщо `a` містить усі елементи `b` |

> 💡 Різниця не симетрична: `a - b` і `b - a` — це різні результати.

Приклад з аналітики: порівнюємо клієнтів, які купували в січні та в лютому.

````python
jan = {"Anna", "Bohdan", "Iryna", "Taras"}     # клієнти, що купували в січні
feb = {"Iryna", "Taras", "Oksana"}             # клієнти, що купували в лютому

print(jan | feb)    # усі клієнти за два місяці
print(jan & feb)    # купували в обох місяцях — лояльні клієнти
print(jan - feb)    # купували в січні, але не повернулися в лютому
print(feb - jan)    # нові клієнти лютого
print(jan ^ feb)    # купували лише в одному з місяців

vip = {"Iryna", "Taras"}
print(vip <= jan)            # True — усі VIP купували в січні
print(vip.issubset(feb))     # True — і в лютому теж
````

## 4. Dictionaries (Словники) — 10 завдань

### 📘 Теорія: що таке словник

**Словник (`dict`)** — колекція пар **ключ → значення**. Замість числового індексу до значення звертаються за ключем — так само, як у справжньому словнику шукають переклад за словом.

```python
client = {"name": "Anna", "age": 28, "city": "Kyiv"}
#          ключ:  значення
```

- **Ключі** унікальні та незмінювані (зазвичай рядки або числа, можна кортежі). Якщо ключ повторити, залишиться останнє значення.
- **Значення** можуть бути будь-якими: числа, рядки, списки, навіть інші словники.
- Словник змінюваний і зберігає порядок, у якому додавалися пари.
- Порожній словник: `{}` або `dict()`.

> 💡 Словник — це по суті **один рядок таблиці**: ключі — назви колонок, значення — дані. А список словників — уже ціла таблиця. Саме так часто виглядають дані, що приходять з API (формат JSON), і саме з такої структури згодом зручно будувати таблиці в pandas.

````python
product = {"title": "Laptop", "price": 32000, "in_stock": True}
student = dict(name="Olena", course="Data Analytics")   # інший спосіб створення

print(product)
print(student)          # {'name': 'Olena', 'course': 'Data Analytics'}
print(len(product))     # 3 — кількість пар
````

### 📘 Доступ до значень

- `d[key]` — повертає значення; якщо такого ключа немає — помилка `KeyError`.
- `d.get(key)` — повертає значення або `None`, якщо ключа немає; `d.get(key, default)` — повертає `default` замість `None`.
- `key in d` — перевіряє наявність **ключа** (не значення!). Для пошуку серед значень: `x in d.values()`.

````python
product = {"title": "Laptop", "price": 32000, "in_stock": True}

print(product["price"])               # 32000
# product["color"]  →  KeyError: 'color'
print(product.get("color"))           # None
print(product.get("color", "n/a"))    # n/a

print("price" in product)             # True
print(32000 in product)               # False — in шукає серед ключів
print(32000 in product.values())      # True — а так шукаємо серед значень
````

### 📘 Зміна словника

| Дія | Синтаксис | Примітка |
|---|---|---|
| Додати / змінити | `d[key] = value` | ключ є — значення перезапишеться; ключа немає — додасться нова пара |
| Видалити | `del d[key]` | `KeyError`, якщо ключа немає |
| Видалити й отримати значення | `d.pop(key)` | |
| Обʼєднати | `d1.update(d2)` | змінює `d1` на місці |
| Обʼєднати в новий словник | <code>d1 &#124; d2</code> | оригінали не змінюються |
| Очистити | `d.clear()` | |

> 💡 Якщо при обʼєднанні ключ є в обох словниках, залишається значення з **правого** (другого) словника.

````python
product = {"title": "Laptop", "price": 32000}

product["price"] = 29900        # ключ існує → зміна
product["brand"] = "Lenovo"     # ключа немає → додавання
print(product)   # {'title': 'Laptop', 'price': 29900, 'brand': 'Lenovo'}

old_price = product.pop("price")
print(old_price, product)   # 29900 {'title': 'Laptop', 'brand': 'Lenovo'}

del product["brand"]
print(product)              # {'title': 'Laptop'}

base = {"currency": "UAH", "vat": 20}
extra = {"vat": 0, "discount": 5}
print(base | extra)         # {'currency': 'UAH', 'vat': 0, 'discount': 5}
base.update(extra)
print(base)                 # {'currency': 'UAH', 'vat': 0, 'discount': 5}
````

### 📘 Перегляд словника: `keys()`, `values()`, `items()`

| Метод | Повертає |
|---|---|
| `d.keys()` | усі ключі |
| `d.values()` | усі значення |
| `d.items()` | пари `(ключ, значення)` у вигляді кортежів |

Ці методи повертають особливі «перегляди» словника. Щоб побачити їх як звичайний список, обгорніть у `list()`. А `values()` чудово працює з `sum()`, `min()`, `max()`, `len()` — так можна порахувати суму чи середнє значення без жодного перебору вручну.

> 💡 Кожен елемент `items()` — це кортеж `(ключ, значення)`, тому його можна розпакувати (як у розділі про кортежі).

````python
stock = {"apples": 40, "pears": 15, "plums": 25}

print(list(stock.keys()))     # ['apples', 'pears', 'plums']
print(list(stock.values()))   # [40, 15, 25]
print(list(stock.items()))    # [('apples', 40), ('pears', 15), ('plums', 25)]

pairs = list(stock.items())
item, qty = pairs[0]          # розпаковуємо першу пару
print(item, "->", qty)        # apples -> 40

print("Всього на складі:", sum(stock.values()))    # 80
print("Позицій:", len(stock))                       # 3
print("Середній залишок:", sum(stock.values()) / len(stock))   # 26.666666666666668
print("Найбільший залишок:", max(stock.values()))  # 40
````

### 📘 Вкладені структури: список словників = таблиця

Колекції можна вкладати одна в одну. Найпоширеніший для аналітика варіант — список словників, де кожен словник є рядком таблиці.

````python
orders = [
    {"id": 1, "client": "Anna",   "amount": 1200},
    {"id": 2, "client": "Bohdan", "amount": 450},
    {"id": 3, "client": "Anna",   "amount": 800},
]

print(orders[0]["client"])     # Anna — перший рядок «таблиці», колонка client
print(len(orders))             # 3 — кількість рядків

print(orders[2]["amount"])     # 800 — третій рядок, колонка amount

orders[1]["amount"] = 500      # змінюємо значення всередині «таблиці»
print(orders[1])               # {'id': 2, 'client': 'Bohdan', 'amount': 500}
````

## 🧭 Підсумок: яку колекцію обрати?

- Потрібна впорядкована послідовність, яку будете змінювати → **`list`**
- Дані не повинні змінюватися (запис, координати, ключ словника) → **`tuple`**
- Важлива унікальність або треба порівняти два набори (спільне, відмінне) → **`set`**
- Треба знаходити значення за назвою чи ідентифікатором, описати обʼєкт з полями → **`dict`**

| | `list` | `tuple` | `set` | `dict` |
|---|---|---|---|---|
| Синтаксис | `[1, 2]` | `(1, 2)` | `{1, 2}` | `{"k": 1}` |
| Порожня колекція | `[]` | `()` | `set()` | `{}` |
| Доступ до елемента | за індексом | за індексом | немає (лише `in`) | за ключем |
| Змінюваний | ✅ | ❌ | ✅ | ✅ |
| Дублікати | ✅ | ✅ | ❌ | ключі — ❌ |
| Перетворення | `list(x)` | `tuple(x)` | `set(x)` | `dict(...)` |
