1. Створи змінну `aqi = 85` — індекс якості повітря. Виведи оцінку якості повітря:

- від 51 до 100 — `"Moderate"`;
- від 101 до 150 — `"Unhealthy for sensitive groups"`;
- від 151 до 200 — `"Unhealthy"`;
- понад 200 — `"Hazardous"`.

Перевір програму для значень 30, 85, 150 і 240

Для підтримуваних валют переведи суму в гривні та розрахуй комісію: якщо сума в гривнях від 20 000 грн — 0.5%, інакше — 1%. Виведи суму в гривнях, комісію та суму до видачі (сума в гривнях мінус комісія).

Перевір програму для пар (`"EUR"`, 500), (`"USD"`, 1000), (`"PLN"`, 300) і (`"GBP"`, 100)

```python
# Початкові дані
currency = "EUR"
amount = 500

# Курси валют
exchange_rates = {
    "USD": 41.2,
    "EUR": 44.8,
    "PLN": 10.3
}

# Перевіряємо, чи є валюта у словнику
if currency in exchange_rates:

    # Переводимо суму в гривні
    amount_in_uah = amount * exchange_rates[currency]

    # Розраховуємо комісію
    if amount_in_uah >= 20000:
        exchange_fee = amount_in_uah * 0.005
    else:
        exchange_fee = amount_in_uah * 0.01

    # Розраховуємо суму до видачі
    total_amount = amount_in_uah - exchange_fee

    # Виводимо результати
    print(f"Сума в гривнях: {amount_in_uah:.2f}")
    print(f"Комісія: {exchange_fee:.2f}")
    print(f"Сума до видачі: {total_amount:.2f}")

else:
    print("Currency not supported")
```

5. Створи змінні `a = 3`, `b = 4`, `c = 5` — довжини сторін трикутника. Виведи тип трикутника:
- якщо сума будь-яких двох сторін не більша за третю, такого трикутника не існує — `"Not a triangle"`;
- усі сторони рівні — `"Equilateral"`;
- дві сторони рівні — `"Isosceles"`;
- трикутник прямокутний (квадрат однієї сторони дорівнює сумі квадратів двох інших) — `"Right"`;
- в інших випадках — `"Scalene"`.

Перевір програму для трійок (3, 4, 5), (5, 3, 4), (2, 2, 2), (5, 5, 8), (1, 2, 3) і (4, 6, 7)

```python
# Сторони трикутника
a = 3
b = 4
c = 5

# Перевіряємо, чи існує трикутник
if a + b <= c or a + c <= b or b + c <= a:
    print("Not a triangle")

# Усі три сторони рівні
elif a == b == c:
    print("Equilateral")

# Будь-які дві сторони рівні
elif a == b or a == c or b == c:
    print("Isosceles")

# Перевіряємо, чи трикутник прямокутний
elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
    print("Right")

# Усі інші випадки
else:
    print("Scalene")
```

6. Виведи зворотний відлік від 10 до 1, кожне число з нового рядка, а після нього — `"Liftoff!"`

7. Створи змінну `number = 90517`. Порахуй суму цифр числа та кількість цифр у ньому.  
Не перетворюй число на рядок. Перевір програму також для числа 7

```python
# Наше число
number = 90517

# Змінна для суми цифр
digit_sum = 0

# Змінна для кількості цифр
digit_count = 0

# Повторюємо, поки число більше нуля
while number > 0:

    # Отримуємо останню цифру
    digit = number % 10

    # Додаємо цифру до суми
    digit_sum = digit_sum + digit

    # Збільшуємо кількість цифр на 1
    digit_count = digit_count + 1

    # Прибираємо останню цифру числа
    number = number // 10

# Виводимо результат
print("Сума цифр:", digit_sum)
print("Кількість цифр:", digit_count)
```

8. Кредит на 30 000 грн. На початку кожного місяця на залишок боргу нараховується 1%, після чого ти вносиш платіж 3000 грн. Останній платіж може бути меншим — рівно стільки, скільки залишилося сплатити.

Для кожного місяця виведи номер місяця й залишок боргу після платежу (округли до 2 знаків). Наприкінці виведи, за скільки місяців кредит буде повністю погашено, і суму останнього платежу

```python
debt = 30000
interest_rate = 0.01
monthly_payment = 3000
month = 0
last_payment = 0

while debt > 0:
    month += 1

    # Нараховуємо відсотки
    debt = round(debt * (1 + interest_rate), 2)

    # Визначаємо розмір платежу
    last_payment = min(monthly_payment, debt)

    # Віднімаємо платіж
    debt = round(debt - last_payment, 2)

    # Виводимо залишок боргу
    print(f"Місяць {month}: залишок боргу {debt:.2f} грн")

print(f"Кредит погашено за {month} місяців")
print(f"Останній платіж: {last_payment:.2f} грн")
```

9. Гіпотеза Колатца. Створи змінну `n = 27` і повторюй дії, доки `n` не стане рівним 1:
- якщо `n` парне — поділи його на 2;
- якщо непарне — помнож на 3 і додай 1.

Виведи, скільки кроків знадобилося, щоб дійти до 1, і яке найбільше значення `n` трапилося по дорозі.  
Перевір програму також для `n = 6`

```python
n = 27
steps = 0
max_number = n

while n != 1:

    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1

    steps += 1

    if n > max_number:
        max_number = n

print(f"Кількість кроків: {steps}")
print(f"Найбільше значення: {max_number}")
```

10. Реєстрація пароля.

**Крок 1.** Запитуй у користувача пароль доти, доки він не відповідатиме всім вимогам. Якщо пароль не підходить, виводь повідомлення про **першу** невиконану вимогу:
- не менше 8 символів — інакше `"Password is too short"`;
- без пробілів — інакше `"Password must not contain spaces"`;
- містить хоча б одну цифру — інакше `"Password must contain a digit"`.

**Крок 2.** Попроси повторити пароль. У користувача 3 спроби: при збігу виведи `"Registration successful"`, при незбігу — `"Passwords do not match. Attempts left:"` і кількість спроб, що залишилися. Якщо спроби закінчилися — виведи `"Registration failed"`.

Перевір програму, ввівши по черзі: `short1`, `long password1`, `longpassword`, `secure123`, `secure12`, `secure123`

```python
# Крок 1. Створення пароля

password = input("Створіть пароль: ")

while True:

    if len(password) < 8:
        print("Password is too short")

    elif " " in password:
        print("Password must not contain spaces")

    elif not any(char.isdigit() for char in password):
        print("Password must contain a digit")

    else:
        break

    password = input("Створіть пароль ще раз: ")


# Крок 2. Підтвердження пароля

attempts_left = 3

while attempts_left > 0:

    confirm_password = input("Повторіть пароль: ")

    if confirm_password == password:
        print("Registration successful")
        break

    else:
        attempts_left -= 1
        print(f"Passwords do not match. Attempts left: {attempts_left}")

        if attempts_left == 0:
            print("Registration failed")
```
