## 1. Dictionaries (Словники) — 5 завдань

1. Створи словник `menu` для кавʼярні з такими напоями та цінами:

| Напій | Ціна, грн |
|---|---|
| espresso | 45 |
| americano | 55 |
| cappuccino | 70 |
| latte | 75 |
| cocoa | 65 |

Назва напою — ключ, ціна — значення

````python
menu = {
    "espresso": 45,
    "americano": 55,
    "cappuccino": 70,
    "latte": 75,
    "cocoa": 65
}

print(menu)
````

2. Виведи ціну напою `"latte"`.  
Потім виведи ціну напою `"raf"` так, щоб програма не завершилася помилкою, а вивела `"not on the menu"`

````python
print("Ціна latte:", menu["latte"])

print("Ціна raf:", menu.get("raf", "not on the menu"))
````

3. Внеси зміни в меню:
- додай напій `"flat white"` за ціною 80 грн;
- збільш ціну `"americano"` на 5 грн;
- прибери `"cocoa"` з меню, зберігши його ціну у змінну `cocoa_price`.

Виведи `menu` і `cocoa_price`

````python
menu["flat white"] = 80

menu["americano"] += 5

cocoa_price = menu.pop("cocoa")

print("Оновлене меню:", menu)
print("Ціна cocoa:", cocoa_price)
````

4. Кава подорожчала: збільш ціни `"cappuccino"` і `"latte"` на 8%, округливши їх до 2 знаків після коми.  
Виведи найвищу, найнижчу та середню ціну в меню (середню округли до 2 знаків)

````python
menu["cappuccino"] = round(menu["cappuccino"] * 1.08, 2)
menu["latte"] = round(menu["latte"] * 1.08, 2)

print("Оновлене меню:", menu)

print("Найвища ціна:", max(menu.values()))
print("Найнижча ціна:", min(menu.values()))
print("Середня ціна:", round(sum(menu.values()) / len(menu), 2))
````

5. Кавʼярня запускає сезонне меню:  
`seasonal = {"pumpkin latte": 95, "latte": 85, "mulled wine": 110}`

Створи словник `full_menu`, який містить усі позиції з `menu` і `seasonal`. Якщо напій є в обох словниках, має залишитися ціна із сезонного меню. Сам `menu` при цьому змінюватися не повинен.

Виведи кількість позицій у `menu` і в `full_menu`, а також перевір, чи є в `full_menu` напій за ціною 110 грн і чи є в ньому `"cocoa"`

````python
seasonal = {
    "pumpkin latte": 95,
    "latte": 85,
    "mulled wine": 110
}

full_menu = {**menu, **seasonal}

print("Повне меню:", full_menu)

print("Кількість позицій у menu:", len(menu))
print("Кількість позицій у full_menu:", len(full_menu))

print("Чи є напій за 110 грн:", 110 in full_menu.values())
print("Чи є cocoa:", "cocoa" in full_menu)
````

12. Код нижче має:
- вивести кількість відвідувачів у четвер, а якщо даних немає — `0`;
- вивести найбільшу кількість відвідувачів за день;
- перевірити, чи був день, коли прийшло 95 відвідувачів.

Знайди й виправ помилки, поясни кожну в коментарі

````python
# visits = {"Mon": 120, "Tue": 95, "Wed": 140}
# print(visits["Thu"])
# print(max(visits))
# print(95 in visits)

visits = {"Mon": 120, "Tue": 95, "Wed": 140}

# Помилка 1: visits["Thu"] викликає KeyError,
# тому що ключа "Thu" немає у словнику.
# Метод get() повертає 0, якщо ключ відсутній.
print(visits.get("Thu", 0))

# Помилка 2: max(visits) знаходить найбільший ключ,
# а не найбільшу кількість відвідувачів.
# Метод values() дозволяє порівнювати значення.
print(max(visits.values()))

# Помилка 3: 95 in visits перевіряє ключі,
# а не значення словника.
# Для перевірки значень використовуємо values().
print(95 in visits.values())
````

### 📚 Міні-задача: облік бібліотеки

Дані:
- `books` — книги в наявності: назва → кількість примірників;
- `borrowed_today` — книги, видані сьогодні: кожен елемент — один виданий примірник;
- `readers_mon`, `readers_tue` — читачі, які приходили в понеділок і у вівторок.

Запусти клітинку з даними, а потім виконуй завдання 14 і 15

````python
books = {"1984": 3, "Dune": 2, "Atomic Habits": 4, "Sapiens": 1}
borrowed_today = ["Dune", "1984", "Sapiens", "1984"]
readers_mon = {"Iryna", "Maksym", "Oleh"}
readers_tue = {"Maksym", "Sofia"}
````

14. Збережи стан `books` до видачі у змінну `books_before` так, щоб подальші зміни `books` на неї не вплинули.  
Зменш кількість примірників кожної книги на кількість виданих сьогодні.  
Виведи `books_before` і `books`, а також множину книг, які сьогодні не видавалися

````python
books_before = books.copy()

for book in borrowed_today:
    books[book] -= 1

not_borrowed = set(books) - set(borrowed_today)

print("Книги до видачі:", books_before)
print("Книги після видачі:", books)
print("Книги, які не видавалися:", not_borrowed)
````

15. Збери звіт у словник `report` з ключами:
- `"titles"` — скільки різних назв книг є в бібліотеці;
- `"copies_left"` — скільки примірників залишилося загалом;
- `"out_of_stock"` — чи є книга, якої не залишилося в наявності (`True` / `False`);
- `"unique_readers"` — скільки різних читачів приходило за два дні;
- `"loyal_readers"` — множина читачів, які приходили обидва дні.

Виведи `report`

````python
report = {
    "titles": len(books),
    "copies_left": sum(books.values()),
    "out_of_stock": 0 in books.values(),
    "unique_readers": len(readers_mon | readers_tue),
    "loyal_readers": readers_mon & readers_tue
}

print(report)
````
